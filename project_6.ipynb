
## This notebook defines functions for sin and cos and tabulates them over a specific range.

```python
import math

def get_sin(x):
    result = math.sin(x)
    return result

def get_cos(x):
    result = math.cos(x)
    return result
```

The get_sin function takes a number and gives back its sine. The get_cos function takes a number and gives back its cosine.

```python
x_values = []
sin_values = []
cos_values = []

for i in range(1000):
    x = 0 + (i*(2/999))
    x_values.append(x)
    sin_values.append(get_sin(x))
    cos_values.append(get_cos(x))
```

```python
for i in range(10):
    print(f"{x_values[i]:.4f}, {sin_values[i]:.4f}, {cos_values[i]:.4f}")
```
