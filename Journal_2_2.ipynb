```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(0, 2 * np.pi, 100)
a, b = 5.0, 0.5
y_clean = a * np.exp(-b * x)

sigma = np.linspace(2, 0.2, 100)
y_scatter = np.random.normal(y_clean, sigma)
```

```python
plt.errorbar(x, y_scatter, yerr=sigma, fmt='o', color='black')
plt.xlabel('x')
plt.ylabel('y')
plt.savefig('scatter_decay_plot.pdf')
plt.show()
```
