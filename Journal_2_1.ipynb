```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(0, 2 * np.pi, 100)

def decay_function(x, a, b):
    return a * np.exp(-b * x)

a = 5.0
b = 0.5
y = decay_function(x, a, b)
```

```python
plt.figure(figsize=(8, 5))
plt.plot(x, y, label=f'y = {a}*exp(-{b}*x)')

plt.xlabel('x (0 to 2π)')
plt.ylabel('y')
plt.title('Exponential Decay Plot')
plt.savefig('exponential_decay_plot.pdf')
plt.show()
```
